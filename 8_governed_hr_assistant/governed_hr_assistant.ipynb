# Project 8: Governed HR assistant

## 1. What can happen?

An assistant can pass one check while missing evidence, carrying an old approval, or being suspended after an incident.

**Example** The prompt changes but the selected evaluation and approval still describe the baseline artifacts.

## 2. Our idea

We combine the earlier small checks around a local Python responder. Four fictional cases use the same question. Sample risk acceptance and approval are clearly fictional; the learner's separate approval template is pending. This capstone blocks confirmed privacy incidents; the explicit recovery workflow is studied in Project 7.

### How we do it

1. Inspect the shared fictional application records and four cases.
2. Run inventory, risk, data, evaluation, artifact-binding, approval, and incident checks before the responder.
3. Select existing current evidence for one case and compare the same requests.
4. Save local decision reports with full version identities and remaining work.

## 3. How it helps

Only eligible cases reach the local responder. The logs prove the boundary ran. Passing fixtures do not establish real approvals, authenticated identities, or compliance. Risk concerns remain visible in the source records even when a fictional reviewer accepted them.

### Step 1: Check the lesson folder

This cell only checks the working directory and imports standard-library tools.

```python
import json
from copy import deepcopy
from datetime import date
from pathlib import Path

project_dir = Path.cwd()
assert project_dir.name == "8_governed_hr_assistant", "Open this numbered folder in VS Code."
as_of = date(2026, 11, 15)
print("Lesson:", project_dir.name, "| fixed sample date:", as_of)
```

### Step 2: Read the shared records and cases

All reports, risk acceptances, and approvals here are fictional. The saved artifact bindings are fixed fixtures, not regenerated during the lesson.

```python
base = json.loads((project_dir / "docs/dataset/base.json").read_text())
scenarios = json.loads((project_dir / "docs/dataset/cases.json").read_text())
results_dir = project_dir / "docs/evalresults"
print(json.dumps(scenarios, indent=2))
print("Sample approval provenance:", base["approval"]["provenance"])
```

### Step 3: Run the enforced baseline

Read tools/governance.py and its small local helper files. review_case collects separate check results; run_cases calls models/demo.py only when every check passes. show_results prints decisions and actual invocation counts. These modules are local copies, so this project has no imports from sibling lesson folders.

```python
from tools.governance import run_cases, show_results

baseline, baseline_calls = run_cases(base, scenarios, project_dir, as_of)
show_results(baseline)
print("Actual local calls:", len(baseline_calls))
```

### Step 4: Inspect why cases stopped

The changed prompt must fail artifact binding even though its release ID is unchanged. The suspended case remains blocked even though its incident record says resolved.

```python
for result in baseline:
    failed = [check["stage"] for check in result["checks"] if not check["passed"]]
    print(result["case_id"], "failed stages:", failed)
print("Documented residual risk:", base["risks"][0]["residual_concern"])
```

### Step 5: Select one existing evidence report

Change only missing_evidence's report selection. Keep the prompt, policy, approvals, risks, and incident cases fixed. No new results or human decisions are manufactured.

```python
candidate_scenarios = deepcopy(scenarios)
candidate_scenarios[1]["report_id"] = "current"
candidate, candidate_calls = run_cases(base, candidate_scenarios, project_dir, as_of)
show_results(candidate)
print("Actual local calls:", len(baseline_calls), "->", len(candidate_calls))
```

Read tools/reports.py. save_report writes uniquely named JSON and Markdown reports in docs/evalresults.
