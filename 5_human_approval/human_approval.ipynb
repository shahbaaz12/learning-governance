# Project 5: Human approval

## 1. What can happen?

A passing report does not tell us whether an authorized person approved this release.

**Example** A developer submits a sample approval that claims an HR reviewer role.

## 2. Our idea

We load a fictional reviewer table separately from six sample approval records. The example policy requires an active release reviewer who is different from the author. Local files demonstrate the checks; they do not authenticate people.

### How we do it

1. Compare pending, rejected, unauthorized, wrong-version, approved, and inactive-reviewer fixtures.
2. Use authority from the separate reviewer table; ignore a role claimed inside an approval.
3. Change only the approved sample's release ID and compare. Read your own pending template separately.

## 3. How it helps

Only the authorized sample for this exact release is approved. Changing its release ID blocks it. Technical evidence is a separate prerequisite, and no deployment occurs here.

### Step 1: Check the lesson folder

This cell only checks the working directory and imports standard-library tools.

```python
import json
from copy import deepcopy
from datetime import date
from pathlib import Path

project_dir = Path.cwd()
assert project_dir.name == "5_human_approval", "Open this numbered folder in VS Code."
as_of = date(2026, 11, 15)
print("Lesson:", project_dir.name, "| fixed sample date:", as_of)
```

### Step 2: Read separate authority and sample decisions

The roles in reviewers.json stand for authenticated application state. A role or admin claim inside an approval is not trusted.

```python
release = json.loads((project_dir / "docs/dataset/release.json").read_text())
reviewers = json.loads((project_dir / "docs/dataset/reviewers.json").read_text())
approvals = json.loads((project_dir / "docs/dataset/approvals.json").read_text())
print(json.dumps(approvals, indent=2))
```

### Step 3: Check the six samples

Read tools/approvals.py. check_approval checks authority and exact release identity before accepting an explicit sample decision. check_cases repeats it and show_results displays reasons.

```python
from tools.approvals import check_cases, show_results

baseline = check_cases(release, approvals, reviewers)
show_results(baseline)
```

### Step 4: Change one release ID

Copy the fixtures and change only the previously approved sample. Keep the same reviewer and explicit decision.

```python
candidate_approvals = deepcopy(approvals)
candidate_approvals["approved_fixture"]["release_id"] = "hr-v2"
candidate = check_cases(release, candidate_approvals, reviewers)
show_results(candidate)
```

### Step 5: Compare approval decisions

An approval for another release must not carry over.

```python
for name in baseline:
    print(name, baseline[name]["decision"], "->", candidate[name]["decision"])
```

### Step 6: Read your own unfilled record

This template is separate from every approved fixture above. The notebook records no human action and writes no approval.

```python
learner_approval = json.loads((project_dir / "docs/dataset/learner_approval.json").read_text())
print(json.dumps(learner_approval, indent=2))
```
